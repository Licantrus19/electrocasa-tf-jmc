```python
%sql

CREATE OR REPLACE FUNCTION electrocasa_dev.silver.mask_dni(valor STRING)
RETURN
  CASE
    WHEN is_account_group_member('electrocasa_engineers') THEN valor
    ELSE '********'
  END;

CREATE OR REPLACE FUNCTION electrocasa_dev.silver.mask_salario(valor DOUBLE)
RETURN
  CASE
    WHEN is_account_group_member('electrocasa_engineers') THEN valor
    ELSE NULL
  END;

ALTER TABLE electrocasa_dev.silver.empleados_historial
ALTER COLUMN dni
SET MASK electrocasa_dev.silver.mask_dni;

ALTER TABLE electrocasa_dev.silver.empleados_historial
ALTER COLUMN salario
SET MASK electrocasa_dev.silver.mask_salario;
```
