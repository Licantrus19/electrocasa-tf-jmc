```python
schema_ventas = "/Volumes/electrocasa_dev/bronze/landing/schemas/ventas"

ventas = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option("header", "true")
        .option("cloudFiles.schemaLocation", schema_ventas)
        .option("rescuedDataColumn", "_rescued_data")
        .load(ruta)
)

ventas.printSchema()
```
