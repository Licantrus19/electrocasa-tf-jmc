```python
%sql

SELECT
    timestamp,
    event_type,
    level,
    message
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0') --electrocasa_pipeline id 
WHERE level IN ('INFO', 'WARN', 'ERROR')
ORDER BY timestamp DESC
LIMIT 50;
```

```python
%sql

SELECT
    level,
    event_type,
    COUNT(*) AS cantidad_eventos
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
GROUP BY
    level,
    event_type
ORDER BY
    level,
    cantidad_eventos DESC;
```

```python
%sql

SELECT
    timestamp,
    level,
    event_type,
    message
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level IN ('WARN', 'ERROR')
ORDER BY timestamp DESC;
```

```python
%sql

SELECT
    timestamp,
    message,
    details
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR'
ORDER BY timestamp DESC;
```

```python
%sql

SELECT
    timestamp,
    message,
    get_json_object(details, '$.update_progress.state') AS estado,
    get_json_object(details, '$.update_progress.cause') AS causa
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR'
ORDER BY timestamp DESC;
```

```python
%sql

SELECT DISTINCT
    explode(json_object_keys(details)) AS campo
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR';
```

```python
%sql

SELECT DISTINCT
    explode(
        json_object_keys(
            get_json_object(details, '$.update_progress')
        )
    ) AS campo
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR';
```

```python
%sql

SELECT
    timestamp,
    message,
    error
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR'
ORDER BY timestamp DESC;
```

```python
%sql

SELECT
    timestamp,
    message,
    error.exceptions[0].class_name AS clase_error,
    error.exceptions[0].message AS detalle_error
FROM event_log('7423e103-e5c9-49aa-9fa1-618bcde703e0')
WHERE level = 'ERROR'
ORDER BY timestamp DESC;
```
