```python
%sql

SELECT
    (SELECT COUNT(*)
     FROM electrocasa_dev.gold.ventas_sucursal_mes) AS filas_gold,

    (SELECT SUM(cantidad_ventas)
     FROM electrocasa_dev.gold.ventas_sucursal_mes) AS ventas_gold,

    (SELECT COUNT(DISTINCT venta_id)
     FROM electrocasa_dev.silver.ventas) AS ventas_silver,

    (SELECT ROUND(SUM(monto_ventas), 2)
     FROM electrocasa_dev.gold.ventas_sucursal_mes) AS monto_gold,

    (SELECT ROUND(SUM(monto_total), 2)
     FROM electrocasa_dev.silver.ventas) AS monto_silver;
```

```python
%sql

SELECT
    (SELECT COUNT(*)
     FROM electrocasa_dev.gold.productos_ventas_devoluciones) AS productos_gold,

    (SELECT SUM(unidades_vendidas)
     FROM electrocasa_dev.gold.productos_ventas_devoluciones) AS unidades_gold,

    (SELECT SUM(cantidad)
     FROM electrocasa_dev.silver.ventas) AS unidades_silver,

    (SELECT SUM(cantidad_devoluciones)
     FROM electrocasa_dev.gold.productos_ventas_devoluciones) AS devoluciones_gold,

    (SELECT COUNT(DISTINCT devolucion_id)
     FROM electrocasa_dev.silver.devoluciones) AS devoluciones_silver;
```
